## 1. query — компактный синтаксис фильтрации
`DataFrame.query()` принимает строку с условием: `df.query('price > 100 and qty >= 2')`. Внутри `query` допускаются `and` и `or`, поскольку строковое выражение анализирует pandas, а не обычный Python. Для передачи значений из внешних переменных используйте `@name`: `df.query('price >= @limit')`. Не собирайте выражение вставкой непроверенного пользовательского текста: это хуже для безопасности и отладки.

````python
import pandas as pd
inventory = pd.DataFrame({'item':['монитор','мышь','кабель'],'stock':[3,0,20],'price':[9000,1500,400]})
limit = 1000
print(inventory.query('stock > 0 and price >= @limit'))
````

## 2. Строковый аксессор .str
Столбцы с текстом обрабатывают через `.str`, что применяет операцию ко всей Series: `str.contains`, `str.startswith`, `str.endswith`, `str.lower`, `str.strip`. В `contains` параметр `regex=False` означает поиск **буквального текста**, а `case=False` — без учёта регистра. Обязательно учитывайте пропуски: `na=False` позволяет получить безопасную логическую маску.

````python
people = pd.DataFrame({'name':['  Анна','борис',None,'Антон'], 'role':['author','reader','author','reader']})
print(people.loc[people['name'].str.contains('ан',case=False,na=False)])
````

## 3. Стабильная сортировка и порядок критериев
`sort_values(['category','price'],ascending=[True,False])` сначала упорядочивает группы по `category`, затем внутри группы — по убыванию цены. `na_position='last'` позволяет оставить пропуски в конце. Если критически важен порядок элементов с равными ключами, помните о стабильности сортировки (`kind='stable'` для одного ключа); явно добавляйте ключ-идентификатор для детерминированного результата.

````python
music = pd.DataFrame({'genre':['jazz','rock','jazz','rock'], 'plays':[4,8,12,3]})
print(music.sort_values(['genre','plays'],ascending=[True,False]))
````

## 4. Критерии выбора инструмента
`loc` с маской универсален и прост в отладке; `query` полезен для читабельных арифметических условий. `.str` нужен для текста. Часто правильное решение комбинирует эти методы: сначала строит строковую маску, затем применяет число условий, после сортирует.
