## Сценарий проекта
Вам предоставлены результаты экспериментов по трём стратегиям обработки состава: последовательной (`seq`), параллельной (`parallel`) и адаптивной (`adaptive`). Для каждого измерения известны стратегия, длительность `time_s`, успешность `success`, нагрузка `cpu`, число обращений `requests` и продукт `product`. Цель — подготовить объективные данные для решения, а не объявить победителя по одному красивому числу.

## 1. Определение метрик до расчёта
Медианное время обработки — `median(time_s)` по всем измерениям стратегии, включая неуспешные, если договорённость исследования не требует иного. **Success rate** — доля `True` от всех запусков стратегии. p90 времени — 90-й процентиль; его нельзя трактовать как гарантию верхнего предела. CPU — отдельный ресурсный показатель, и ускорение за счёт увеличения нагрузки должно отражаться в отчёте.

````python
import pandas as pd
runs=pd.DataFrame({'strategy':['A','A','B'],'time_s':[5.,9.,7.],'success':[True,False,True]})
print(runs.groupby('strategy').agg(median_s=('time_s','median'),success_rate=('success','mean')))
````

## 2. Сравнения и контроль контекста
Если сравнивать стратегии на разных продуктах, результат может отражать сложность продукта, а не качество оркестрации. Поэтому часть анализа делаем по паре `strategy,product`. В полноценном исследовании также фиксируйте устройство, размер датасета, тип этикетки, условия освещения, доверительные интервалы и процедуру повторения. Наш учебный набор иллюстрирует обработку и не доказывает статистически преимуществ какой-либо стратегии.

## 3. Эффект и ресурсы
Для эталона `seq` относительное ускорение времени на стратегии X можно оценить `(median_seq - median_X) / median_seq * 100`. Положительный результат означает меньшее время, отрицательный — медленнее. При нулевой медиане seq знаменатель не определён. Обязательно выводите изменения успешности и CPU рядом с ускорением: одно ускорение не отражает весь trade-off.

## 4. Аналитический вывод
После построения таблицы результатов отдельно запишите, какие выводы подтверждены измерениями и какие требуют большего набора. Не переносите выводы из учебных синтетических данных на реальную систему без проверки повторяемости, сопоставимости условий и статистической неопределённости.
