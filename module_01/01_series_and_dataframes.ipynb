## 1. Зачем нужен pandas

**pandas** — библиотека для анализа табличных данных. Таблица в pandas — не набор произвольных ячеек, как в Excel, а объект с правилами: у каждого столбца есть имя и тип, у каждой строки есть метка индекса. Это позволяет выполнять одну операцию сразу над целым столбцом. Например, умножение столбца `price` на столбец `quantity` создаёт серию значений, а не требует явного цикла по строкам.

Pandas не заменяет SQL и Python: SQL полезен для отбора данных на стороне базы, Python — для логики приложения, pandas — для обработки и исследования полученных таблиц.

**Важно:** pandas действует преимущественно в оперативной памяти. Для очень больших файлов потребуется специальная стратегия загрузки — об этом будет отдельный модуль.

## 2. Series: одномерные данные

`pd.Series` — последовательность значений **вместе с индексом**. Индекс — это метки, а не обязательно порядковые номера.

```python
import pandas as pd
temperatures = pd.Series([18, 21, 19], index=["пн", "вт", "ср"], name="temp")
print(temperatures)
print(temperatures["вт"])
```

Здесь `name` — имя самой серии, `index` — метки элементов, значения — 18, 21, 19. В отличие от списка, Series умеет выполнять вычисления сразу над всеми числами: `temperatures + 2`. Возвращается **новая** Series; исходная не меняется.

Индекс важен для вычислений: при сложении двух Series pandas **сопоставляет значения по меткам**, а не по их визуальному положению. Если метка есть только в одной из серий, результат для неё будет отсутствующим. Не путайте индекс с физическим номером строки.

## 3. DataFrame: таблица из Series

`DataFrame` — двумерная структура: строки × столбцы. Каждый столбец ведёт себя как Series со **своим** типом данных; все столбцы одной таблицы делят общий индекс.

```python
products = pd.DataFrame({
    "sku": [101, 102, 103],
    "category": ["чай", "кофе", "чай"],
    "price": [120.0, 490.0, 160.0],
    "active": [True, True, False],
})
products
```

У `.shape` два значения `(строки, столбцы)`. `.columns` — названия столбцов. `.index` — индекс. `.dtypes` — тип каждого столбца. `.ndim` — число измерений (у Series — 1, у DataFrame — 2).

**Разница `df["price"]` и `df[["price"]]`:** одинарные скобки дают Series, двойные — DataFrame из одного столбца. Это критично для последующей работы.

## 4. Индексы и выравнивание

Индекс — это идентификатор строки в конкретной таблице, но не обязательно уникальный ключ бизнеса. По умолчанию используется `RangeIndex(0, 1, 2, ...)`. Индекс может быть строковым и даже содержать повторы; последствия повторов разберём в отдельном модуле.

```python
a = pd.Series([100, 200], index=["x", "y"])
b = pd.Series([5, 7], index=["y", "x"])
print(a + b)  # x -> 107, y -> 205
```

Это принцип **выравнивания по индексу**: `x` складывается с `x`. Если ожидаете сложение по позициям, сначала проверьте индексы или явно работайте с массивами NumPy. Не сбрасывайте индекс механически — сперва поймите его роль.

## 5. Создание данных и типы

Таблицы можно создавать из словаря списков или списка словарей. Во втором случае строки могут иметь разные наборы ключей: отсутствующее значение заполнится пропуском.

```python
orders = pd.DataFrame([
    {"order_id": 1, "amount": 99.5},
    {"order_id": 2, "amount": 145.0},
])
print(orders.dtypes)
```

Типы: целые числа, вещественные, логические, строки, даты. Число `"120"` как текст — не то же самое, что число `120`: арифметика может не сработать. Проверяйте `.dtypes` до расчётов. В pandas версии 3 строковые типы могут отображаться иначе, чем в старых материалах: понимание значения типа важнее конкретной подписи dtype.

**Типичная ошибка:** указывать один список значений короче других при создании DataFrame из словаря списков. Длины должны совпадать.

## Дополнительный разбор: индексы, типы и выравнивание
При сложении Series pandas соотносит значения **по меткам**, а не по местам. Это отличается от поведения обычных списков и объясняет многие трудные ошибки: если индексы `[a,b]` и `[b,a]` имеют разный порядок, сумма всё равно сопоставит одинаковые метки. Если совпадения нет, обычно появится пропуск.

Таблица состоит из столбцов разных типов. `int64` — целые, `float64` — вещественные, `bool` — логические; строки в pandas 3.x обычно получают отдельный строковый тип. Полезно сразу проверять `df.dtypes`: неправильный тип влияет на сортировку и арифметику. Думайте об индексе как о метках наблюдений, а о названиях столбцов как о переменных, описывающих наблюдение.

Типичная ошибка новичка — использовать `df['x']` и `df[['x']]` как одинаковые операции. Первая возвращает Series, вторая DataFrame. Это повлияет на размерность результата, методы группировки и дальнейшие соединения. После создания таблицы всегда проверяйте `shape`, `columns`, `index` и `dtypes`.
